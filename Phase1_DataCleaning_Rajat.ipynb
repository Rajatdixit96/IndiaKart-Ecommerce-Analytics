```python
import pandas as pd

orders = pd.read_csv("orders DA.csv")
order_items = pd.read_csv("order_items DA.csv")
customers = pd.read_csv("customers DA.csv")
products = pd.read_csv("products DA.csv")
payments = pd.read_csv("payments DA.csv")
returns = pd.read_csv("returns DA.csv")
inventory = pd.read_csv("inventory DA.csv")
suppliers = pd.read_csv("suppliers DA.csv")

print("All files loaded successfully!")
```

```python
datasets = {
    "Orders": orders,
    "Order Items": order_items,
    "Customers": customers,
    "Products": products,
    "Payments": payments,
    "Returns": returns,
    "Inventory": inventory,
    "Suppliers": suppliers
}

for name, df in datasets.items():
    print("="*50)
    print(name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print(df.dtypes)
    print()
```

```python
for name, df in datasets.items():
    print("="*50)
    print(name)
    print(df.isnull().sum())
    print()
```

```python
for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("Duplicate Rows:", df.duplicated().sum())
    print()
```

```python
# Orders
orders['order_date'] = pd.to_datetime(orders['order_date'], format='mixed', dayfirst=True, errors='coerce')
orders['delivered_date'] = pd.to_datetime(orders['delivered_date'], format='mixed', dayfirst=True, errors='coerce')

# Customers
customers['registration_date'] = pd.to_datetime(customers['registration_date'], format='mixed', dayfirst=True, errors='coerce')
customers['last_login_date'] = pd.to_datetime(customers['last_login_date'], format='mixed', dayfirst=True, errors='coerce')

# Products
products['launch_date'] = pd.to_datetime(products['launch_date'], format='mixed', dayfirst=True, errors='coerce')

# Payments
payments['payment_date'] = pd.to_datetime(payments['payment_date'], format='mixed', dayfirst=True, errors='coerce')
payments['refund_date'] = pd.to_datetime(payments['refund_date'], format='mixed', dayfirst=True, errors='coerce')

# Returns
returns['return_date'] = pd.to_datetime(returns['return_date'], format='mixed', dayfirst=True, errors='coerce')
returns['refund_date'] = pd.to_datetime(returns['refund_date'], format='mixed', dayfirst=True, errors='coerce')

# Inventory
inventory['last_restocked_date'] = pd.to_datetime(inventory['last_restocked_date'], format='mixed', dayfirst=True, errors='coerce')

# Suppliers
suppliers['created_date'] = pd.to_datetime(suppliers['created_date'], format='mixed', dayfirst=True, errors='coerce')

print("All date columns converted successfully!")
```
